# 🤖 Notebook 03: Pemodelan Clustering K-Means UMKM Kota Bandung

Notebook ini melakukan pemodelan K-Means clustering, penentuan kluster optimal, dan evaluasi metrik.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

def cari_dataset():
    for p in ["data/raw/data_umkm_bandung.csv", "../data/raw/data_umkm_bandung.csv", "../../data/raw/data_umkm_bandung.csv"]:
        if os.path.exists(p): return p
    raise FileNotFoundError("Dataset tidak ditemukan!")

path = cari_dataset()
df = pd.read_csv(path, sep=None, engine='python')

df['totalScore'] = pd.to_numeric(df['totalScore'], errors='coerce').fillna(df['totalScore'].median() if 'totalScore' in df else 4.0)
df['reviewsCount'] = pd.to_numeric(df['reviewsCount'], errors='coerce').fillna(0)
df['log_reviewsCount'] = np.log1p(df['reviewsCount'])

X = df[['totalScore', 'log_reviewsCount']]
X_scaled = StandardScaler().fit_transform(X)

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X_scaled)

sil = silhouette_score(X_scaled, df['cluster'])
print(f"K-Means Clustering Complete. Silhouette Score: {sil:.4f}")
display(df.groupby('cluster')[['totalScore', 'reviewsCount']].mean())
```
