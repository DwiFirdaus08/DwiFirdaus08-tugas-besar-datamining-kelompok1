# 🧹 Notebook 02: Preprocessing Data UMKM Kota Bandung

Notebook ini menangani missing values, transformasi logaritmik, dan penskalaan fitur.


```python
import os
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

def cari_dataset():
    for p in ["data/raw/data_umkm_bandung.csv", "../data/raw/data_umkm_bandung.csv", "../../data/raw/data_umkm_bandung.csv"]:
        if os.path.exists(p): return p
    raise FileNotFoundError("Dataset tidak ditemukan!")

path = cari_dataset()
df = pd.read_csv(path, sep=None, engine='python')

df['totalScore'] = pd.to_numeric(df['totalScore'], errors='coerce').fillna(df['totalScore'].median() if 'totalScore' in df else 4.0)
df['reviewsCount'] = pd.to_numeric(df['reviewsCount'], errors='coerce').fillna(0)
df['log_reviewsCount'] = np.log1p(df['reviewsCount'])

scaler = StandardScaler()
scaled = scaler.fit_transform(df[['totalScore', 'log_reviewsCount']])
print("Scaled shape:", scaled.shape)
display(pd.DataFrame(scaled, columns=['totalScore_std', 'log_reviewsCount_std']).head())
```
