# 🧪 Notebook 01: Exploratory Data Analysis (EDA) - UMKM Kota Bandung

Notebook ini digunakan untuk mengeksplorasi karakteristik data UMKM di Kota Bandung.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def cari_dataset():
    for p in ["data/raw/data_umkm_bandung.csv", "../data/raw/data_umkm_bandung.csv", "../../data/raw/data_umkm_bandung.csv"]:
        if os.path.exists(p): return p
    raise FileNotFoundError("Dataset tidak ditemukan!")

path = cari_dataset()
df = pd.read_csv(path, sep=None, engine='python')
print(f"Loaded {len(df):,} rows from {path}")
df.info()
```

```python
# Statistik Deskriptif Fitur Utama
display(df.describe())
```

```python
# Visualisasi Distribusi Rating & Jumlah Ulasan
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
if 'totalScore' in df:
    sns.histplot(df['totalScore'], kde=True, ax=axes[0], color='skyblue')
    axes[0].set_title('Distribusi Rating Tempat (totalScore)')
if 'reviewsCount' in df:
    sns.histplot(df['reviewsCount'], kde=True, ax=axes[1], color='salmon')
    axes[1].set_title('Distribusi Jumlah Ulasan (reviewsCount)')
plt.tight_layout()
plt.show()
```
